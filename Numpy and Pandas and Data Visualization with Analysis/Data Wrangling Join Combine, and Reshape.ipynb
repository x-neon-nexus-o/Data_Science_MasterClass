## Table 8-1. Different join types with the `how` argument

| Option        | Behavior                                                        | Example (Python) | Result |
|---------------|-----------------------------------------------------------------|------------------|--------|
| `how="inner"` | Use only the key combinations observed in both tables           | `pd.merge(df1, df2, how="inner", on="key")` | Intersection of keys |
| `how="left"`  | Use all key combinations found in the left table                 | `pd.merge(df1, df2, how="left", on="key")`  | All keys from left, match from right |
| `how="right"` | Use all key combinations found in the right table                | `pd.merge(df1, df2, how="right", on="key")` | All keys from right, match from left |
| `how="outer"` | Use all key combinations observed in both tables together        | `pd.merge(df1, df2, how="outer", on="key")` | Union of keys |

## Arguments for `pandas.merge`

| Argument      | Description                                                                 | Example (Python) |
|---------------|-----------------------------------------------------------------------------|------------------|
| `left`        | DataFrame to be merged on the left side                                     | `pd.merge(df1, df2, ...)` |
| `right`       | DataFrame to be merged on the right side                                    | `pd.merge(df1, df2, ...)` |
| `how`         | Type of join: `"inner"`, `"outer"`, `"left"`, `"right"`; defaults to `"inner"` | `pd.merge(df1, df2, how="outer")` |
| `on`          | Column names to join on; must exist in both DataFrames. If not specified, uses intersection of column names | `pd.merge(df1, df2, on="id")` |
| `left_on`     | Column(s) in left DataFrame to use as join keys                             | `pd.merge(df1, df2, left_on="id1", right_on="id2")` |
| `right_on`    | Column(s) in right DataFrame to use as join keys                            | Same as above |
| `left_index`  | Use row index in left as join key (supports MultiIndex)                     | `pd.merge(df1, df2, left_index=True, right_on="id")` |
| `right_index` | Use row index in right as join key                                          | `pd.merge(df1, df2, right_index=True, left_on="id")` |
| `sort`        | Sort merged data lexicographically by join keys; default `False`            | `pd.merge(df1, df2, on="id", sort=True)` |
| `suffixes`    | Tuple of suffixes for overlapping column names; default `("_x","_y")`       | `pd.merge(df1, df2, on="id", suffixes=("_left","_right"))` |
| `copy`        | If `False`, avoid copying data in some cases; default always copies         | `pd.merge(df1, df2, copy=False)` |
| `validate`    | Verifies merge type: `"1:1"`, `"1:m"`, `"m:1"`, `"m:m"`                     | `pd.merge(df1, df2, on="id", validate="1:1")` |
| `indicator`   | Adds `_merge` column showing source of each row: `"left_only"`, `"right_only"`, `"both"` | `pd.merge(df1, df2, on="id", indicator=True)` |

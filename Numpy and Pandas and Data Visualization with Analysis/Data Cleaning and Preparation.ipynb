## Table 7-1. NA handling object methods

| Method   | Description                                                                 | Example (Python) | Output |
|----------|-----------------------------------------------------------------------------|------------------|--------|
| `dropna` | Filter axis labels based on whether values for each label have missing data, with varying thresholds for how much missing data to tolerate | `pd.Series([1, np.nan, 3]).dropna()` | `[1.0, 3.0]` |
| `fillna` | Fill in missing data with some value or using an interpolation method such as `"ffill"` or `"bfill"` | `pd.Series([1, np.nan, 3]).fillna(0)` | `[1.0, 0.0, 3.0]` |
| `isna`   | Return Boolean values indicating which values are missing/NA                | `pd.Series([1, np.nan, 3]).isna()` | `[False, True, False]` |
| `notna`  | Negation of `isna`, returns True for non-NA values and False for NA values  | `pd.Series([1, np.nan, 3]).notna()` | `[True, False, True]` |

## Table 7-2. `fillna` function arguments

| Argument | Description                                                                 | Example (Python) | Output |
|----------|-----------------------------------------------------------------------------|------------------|--------|
| `value`  | Scalar value or dictionary-like object to use to fill missing values        | `pd.Series([1, np.nan, 3]).fillna(value=0)` | `[1.0, 0.0, 3.0]` |
| `method` | Interpolation method: `"bfill"` (backward fill) or `"ffill"` (forward fill); default is `None` | `pd.Series([1, np.nan, np.nan, 4]).fillna(method="ffill")` | `[1.0, 1.0, 1.0, 4.0]` |
| `axis`   | Axis to fill on (`"index"` or `"columns"`); default is `"index"`            | `df.fillna(method="ffill", axis="columns")` | Missing values filled column-wise |
| `limit`  | For forward/backward filling, maximum number of consecutive periods to fill | `pd.Series([1, np.nan, np.nan, 4]).fillna(method="ffill", limit=1)` | `[1.0, 1.0, NaN, 4.0]` |

## Common Python String Methods

| Method        | Description                                                                 | Example (Python) | Output |
|---------------|-----------------------------------------------------------------------------|------------------|--------|
| `count`       | Return the number of nonoverlapping occurrences of substring in the string  | `"banana".count("a")` | `3` |
| `endswith`    | Return True if string ends with suffix                                      | `"hello".endswith("lo")` | `True` |
| `startswith`  | Return True if string starts with prefix                                    | `"hello".startswith("he")` | `True` |
| `join`        | Use string as delimiter for concatenating a sequence of other strings       | `",".join(["a","b","c"])` | `"a,b,c"` |
| `index`       | Return starting index of first occurrence of substring; raises `ValueError` if not found | `"hello".index("e")` | `1` |
| `find`        | Return position of first occurrence of substring; returns `-1` if not found | `"hello".find("x")` | `-1` |
| `rfind`       | Return position of last occurrence of substring; returns `-1` if not found  | `"banana".rfind("a")` | `5` |
| `replace`     | Replace occurrences of substring with another string                       | `"hello world".replace("world","Python")` | `"hello Python"` |
| `strip`       | Trim whitespace/newlines on both sides                                     | `"  hello  ".strip()` | `"hello"` |
| `rstrip`      | Trim whitespace/newlines on the right side                                 | `"hello   ".rstrip()` | `"hello"` |
| `lstrip`      | Trim whitespace/newlines on the left side                                  | `"   hello".lstrip()` | `"hello"` |
| `split`       | Break string into list of substrings using delimiter                       | `"a,b,c".split(",")` | `["a","b","c"]` |
| `lower`       | Convert alphabet characters to lowercase                                   | `"HELLO".lower()` | `"hello"` |
| `upper`       | Convert alphabet characters to uppercase                                   | `"hello".upper()` | `"HELLO"` |
| `casefold`    | Convert to lowercase, handling region-specific variations for comparison   | `"ß".casefold()` | `"ss"` |
| `ljust`       | Left justify string, padding right side with spaces or fill character      | `"hi".ljust(5,"-")` | `"hi---"` |
| `rjust`       | Right justify string, padding left side with spaces or fill character      | `"hi".rjust(5,"-")` | `"---hi"` |

## Common `re` (regular expression) methods in Python

| Method      | Description                                                                 | Example (Python) | Output |
|-------------|-----------------------------------------------------------------------------|------------------|--------|
| `findall`   | Return all nonoverlapping matching patterns in a string as a list           | `re.findall(r"\d+", "abc123def45")` | `['123','45']` |
| `finditer`  | Like `findall`, but returns an iterator yielding match objects              | `[m.group() for m in re.finditer(r"\d+", "abc123def45")]` | `['123','45']` |
| `match`     | Match pattern at start of string; returns a match object or `None`          | `re.match(r"\d+", "123abc")` | `<Match '123'>` |
| `search`    | Scan string for match; returns first match object (anywhere in string)      | `re.search(r"\d+", "abc123def")` | `<Match '123'>` |
| `split`     | Break string into pieces at each occurrence of pattern                      | `re.split(r"\d+", "abc123def45ghi")` | `['abc','def','ghi']` |
| `sub`       | Replace all occurrences of pattern with replacement string                  | `re.sub(r"\d+", "#", "abc123def45")` | `"abc#def#"` |
| `subn`      | Like `sub`, but also return number of substitutions made                    | `re.subn(r"\d+", "#", "abc123def45")` | `('abc#def#', 2)` |
